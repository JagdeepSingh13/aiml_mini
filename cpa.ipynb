```python
companies = {
    "TCS":       {"cgpa": 6.0, "package": "3.5 LPA", "branches": ["CSE", "IT", "ECE", "EEE"]},
    "Infosys":   {"cgpa": 6.5, "package": "4 LPA",   "branches": ["CSE", "IT", "ECE"]},
    "Accenture": {"cgpa": 6.5, "package": "4.5 LPA", "branches": ["CSE", "IT", "ECE", "ME", "CE"]},
    "Wipro":     {"cgpa": 6.0, "package": "3.8 LPA", "branches": ["CSE", "IT", "ECE"]},
    "Microsoft": {"cgpa": 8.0, "package": "22 LPA",  "branches": ["CSE", "IT"]},
    "Google":    {"cgpa": 8.5, "package": "25 LPA",  "branches": ["CSE", "IT"]},
}

BRANCHES = ["CSE", "IT", "ECE", "EEE", "ME", "CE"]

MENU = '''
1. View Companies
2. Check Eligibility
3. Resume Tips
4. Interview Tips
5. Exit
You can also type:  cgpa 8.2   |   CSE   |   Google'''
```

```python
def sensor():
    text = input("You: ").strip()
    return text

def actuator(res):
    print(f"Actuator: {res} \n")  
```

```python
class SRA:
    def res(self, text):
        if text == "1":
            return "Companies: " + ", ".join(companies)
        elif text == "2":
            return "Sorry, I have no memory, so I cannot check your eligibility."
        elif text == "3":
            return "Resume Tips: keep it one page, highlight projects, mention internships and skills."
        elif text == "4":
            return "Interview Tips: practise DSA, revise DBMS / OOP / OS, prepare HR questions."
        elif text == "5":
            return "Thank you! Best wishes for your placements."
        else:
            return "Please select a valid option (1-5)." 
```

```python
class MBA:
    def __init__(self):
        self.cgpa = None
        self.branch = None

    def res(self, text):
        if text.lower().startswith("cgpa"):
            try:
                self.cgpa = float(text.split()[-1])
            except ValueError:
                return "Please enter CGPA as a number, e.g. cgpa 8.2"
            return f"CGPA Saved: {self.cgpa}"

        if text.upper() in BRANCHES:
            self.branch = text.upper()
            return f"Branch Saved: {self.branch}"

        for n, i in companies.items():
            if text.lower() == n.lower():
                return f"{n} INFO: Min CGPA: {i['cgpa']}, Branches: {', '.join(i['branches'])}"

        if text == "1":
            return "Companies: " + ", ".join(companies)
        elif text == "2":
            if self.cgpa is None or self.branch is None:
                return "First tell me your CGPA (e.g. cgpa 8.2) and branch (e.g. CSE)."
            eligible = [f"{name}, ({info['package']})" for name, info in companies.items()
                        if self.cgpa >= info['cgpa'] and self.branch in info['branches']]
            if eligible:
                return "You are eligible for: " + ", ".join(eligible)
            return "No eligible companies found."
        elif text == "3":
            return "Resume Tips: keep it one page, highlight projects, mention internships and skills."
        elif text == "4":
            return "Interview Tips: practise DSA, revise DBMS / OOP / OS, prepare HR questions."
        elif text == "5":
            return "Thank you! Best wishes for your placements."
        else:
            return "Please select a valid option (1-5)." 
```

```python
print("===== CAMPUS PLACEMENT ASSISTANT =====")
print("1. Simple Reflex Agent")
print("2. Model-Based Agent")
choice = input("Select agent: ").strip()

if choice == "1":
    agent = SRA()
    print("\nUsing Simple Reflex Agent")
else:
    agent = MBA()
    print("\nUsing Model-Based Agent")

print(MENU, "\n")

while True:
    text = sensor()                    # 1. sense
    reply = agent.res(text)       # 2. decide
    actuator(reply)                    # 3. act
    if text == "5":
        break
```
